```python
import torch
from datasets import load_dataset, Dataset
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, TrainingArguments
from trl import SFTTrainer
from peft import LoraConfig

# 1. تحميل البيانات
dataset = load_dataset("Malikeh1375/medical-question-answering-datasets", "all-processed")

# 2. تنسيقها
formatted_data = []
for item in dataset["train"]:
    if item['input']:
        text = f"### Instruction:\n{item['instruction']}\n\n### Input:\n{item['input']}\n\n### Response:\n{item['output']}"
    else:
        text = f"### Instruction:\n{item['instruction']}\n\n### Response:\n{item['output']}"
    formatted_data.append({"text": text})

formatted_dataset = Dataset.from_list(formatted_data)

# 3. تحميل التوكنايزر والموديل
llama_tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-3B", token="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")  # ← ضع التوكن بتاعك
llama_tokenizer.pad_token = llama_tokenizer.eos_token

quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_quant_type="nf4"
)

llama_model = AutoModelForCausalLM.from_pretrained(
    "meta-llama/Llama-3.2-3B",  # ← نفس اللي استخدمته
    quantization_config=quant_config,
    device_map="auto",
    token="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"
)

# 4. توكنزة البيانات
def tokenize(example):
    return llama_tokenizer(
        example["text"],
        truncation=True,
        padding="max_length",
        max_length=512,
        return_tensors="pt"
    )

tokenized_dataset = formatted_dataset.map(tokenize, batched=True)

# 5. إعدادات التدريب
training_arguments = TrainingArguments(
    output_dir="./llama-medical",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    num_train_epochs=.02,
    learning_rate=2e-4,
    logging_steps=10,
    save_strategy="epoch",
    fp16=True,
    report_to="none"
)

# 6. LoRA config
lora_config = LoraConfig(
    r=64,
    lora_alpha=16,
    lora_dropout=0.1,
    bias="none",
    task_type="CAUSAL_LM"
)

# 7. تدريب الموديل (بدون tokenizer)
trainer = SFTTrainer(
    model=llama_model,
    args=training_arguments,
    train_dataset=tokenized_dataset,
    peft_config=lora_config,
)

trainer.train()
```

```python
trainer.model.save_pretrained("./llama-medical-model")
llama_tokenizer.save_pretrained("./llama-medical-model")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import os

# إنشاء مجلد offload لو مش موجود
os.makedirs("./offload", exist_ok=True)

# تحميل النموذج مع تحديد مجلد offload
model = AutoModelForCausalLM.from_pretrained(
    "./llama-medical-model",
    device_map="auto",
    offload_folder="./offload"
)

# تحميل التوكنيزر
tokenizer = AutoTokenizer.from_pretrained("./llama-medical-model")
```

```python
def chat_with_model(prompt, model, tokenizer):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_new_tokens=200, do_sample=True, temperature=0.7, top_p=0.9)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
import torch
prompt = """### Instruction:
I have a headache and feel dizzy. What should I do?

### Response:
"""

response = chat_with_model(prompt, model, tokenizer)
print(response)
```

```python
# Define a few test dental prompts
dental_prompts = [
    "I have a sharp pain when I eat something cold. What's going on?",
    "My gums are bleeding when I brush my teeth. Should I be worried?",
    "There’s a small hole in my tooth and it hurts. What could it be?",
    "I feel swelling and pain near one of my molars. What should I do?",
    "My tooth cracked when I bit something hard. What's the treatment?"
]

# Loop through the prompts and get responses from the model
for i, prompt in enumerate(dental_prompts, 1):
    full_prompt = f"### Instruction:\n{prompt}\n\n### Response:\n"
    response = chat_with_model(full_prompt, model, tokenizer)
    print(f"\n--- Test Case {i} ---")
    print(f"Prompt: {prompt}")
    print("Response:", response)
```
